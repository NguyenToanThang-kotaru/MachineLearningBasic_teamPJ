```python
import math

# Hàm kích hoạt Sigmoid
def sigmoid(x):
    if x < -709: return 0.0
    if x > 709: return 1.0
    return 1.0 / (1.0 + math.exp(-x))

# Hàm tính Loss
def calculate_loss(y_true, y_pred):
    epsilon = 1e-15
    y_pred = max(min(y_pred, 1 - epsilon), epsilon)
    return -(y_true * math.log(y_pred) + (1 - y_true) * math.log(1 - y_pred))
```

```python
print("--- NHẬP THAM SỐ ---")
num_epoch = int(input("num_epoch: "))
agr_size = int(input("agr_size (số lượng features): "))
epoch_size = int(input("epoch_size (số lượng mẫu dữ liệu): "))
learning_rate = float(input("learning_rate: "))
bias = float(input("bias ban đầu: "))

# Khởi tạo mảng lưu dữ liệu
# X: Danh sách các mẫu, mỗi mẫu là một danh sách các features
# Y: Danh sách nhãn thực tế
X = []
Y = []

print(f"\n--- NHẬP DỮ LIỆU (cho {epoch_size} mẫu) ---")
for j in range(epoch_size):
    print(f"Data {j + 1}:")
    sample_x = []
    for t in range(agr_size):
        val = float(input(f"  Feature {t + 1}: "))
        sample_x.append(val)
    X.append(sample_x)

    val_y = float(input(f"  Label y [{j + 1}]: "))
    Y.append(val_y)

# Nhập trọng số w ban đầu
arr_w = [0.0] * agr_size
print("\n--- NHẬP TRỌNG SỐ W BAN ĐẦU ---")
for i in range(agr_size):
    arr_w[i] = float(input(f"w[{i}]: "))
```

```python
print("\n--- BẮT ĐẦU TRAINING ---")

for i in range(1, num_epoch + 1):
    print(f"\nEpoch {i}: --------------------------------")
    print(f"w hiện tại: {arr_w}")
    print(f"bias hiện tại: {bias}")

    # Mảng lưu gradient tổng hợp cho từng feature
    grad_w_sum = [0.0] * agr_size
    grad_bias_sum = 0.0

    # Duyệt qua từng mẫu dữ liệu
    for j in range(epoch_size):
        # 1. Tính z = bias + w1*x1 + w2*x2 + ...
        z = bias
        for t in range(agr_size):
            z += X[j][t] * arr_w[t]

        # 2. Tính sigmoid(z)
        pred = sigmoid(z)

        # 3. Tính loss
        loss_val = calculate_loss(Y[j], pred)
        print(f"  Data {j+1}: z = {z:.4f}, sigmoid = {pred:.4f}, loss = {loss_val:.4f}")

        # 4. Tính Gradient cho từng mẫu
        # Đạo hàm của Loss theo z là (pred - y)
        error = pred - Y[j]

        # Gradient cho bias
        grad_bias_sum += error

        # Gradient cho từng w[t]
        for t in range(agr_size):
            grad_w_t = error * X[j][t]
            grad_w_sum[t] += grad_w_t
            print(f"    grad_w[{t}]: {grad_w_t:.4f}")

    print("Medium (Average Gradients):")

    # 5. Cập nhật trọng số (Weights Update)
    # w_new = w_old - learning_rate * average_gradient
    for t in range(agr_size):
        avg_grad_w = grad_w_sum[t] / epoch_size
        print(f"  Avg grad_w[{t}]: {avg_grad_w:.4f}")
        arr_w[t] = arr_w[t] - learning_rate * avg_grad_w

    # Cập nhật bias
    avg_grad_bias = grad_bias_sum / epoch_size
    bias = bias - learning_rate * avg_grad_bias

print("\n--------------------------------")
print("KẾT QUẢ CUỐI CÙNG:")
print(f"Final Weights: {arr_w}")
print(f"Final Bias: {bias}")
```
